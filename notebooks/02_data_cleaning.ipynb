```python
import pandas as pd
import numpy as np
file_path = "../data/raw/sales_data.csv"
df = pd.read_csv(file_path)
print("Shape:", df.shape)
df.head()
```

```python
df.columns = (
    df.columns
    .str.strip()
    .str.lower()
    .str.replace(" ", "_")
)
print(df.columns.tolist())
```

```python
df["date"] = pd.to_datetime(
    df["date"],
    errors="coerce"
)
print(df["date"].dtype)
```

```python
print("Invalid dates:", df["date"].isna().sum())
```

```python
missing = df.isnull().sum()
print(missing[missing > 0])
```

```python
missing_percentage = (
    df.isnull().mean() * 100
).sort_values(ascending=False)
print(missing_percentage[missing_percentage > 0])
```

```python
duplicates = df.duplicated().sum()
print("Duplicate rows:", duplicates)
```

```python
numeric_columns = df.select_dtypes(
    include=np.number
).columns
print(numeric_columns.tolist())
```

```python
df[numeric_columns].describe().T
```

```python
print(
    "Negative units sold:",
    (df["units_sold"] < 0).sum()
)
```

```python
print(
    "Invalid prices:",
    (df["price"] <= 0).sum()
)
```

```python
print(
    "Negative inventory:",
    (df["inventory_level"] < 0).sum()
)
```

```python
print(
    "Negative demand:",
    (df["demand"] < 0).sum()
)
```

```python
print("Products:", df["product_id"].nunique())
print("Stores:", df["store_id"].nunique())
print("Categories:", df["category"].nunique())
print("Regions:", df["region"].nunique())
```

```python
print("\nCategories:")
print(df["category"].unique())
```

```python
print("\nRegions:")
print(df["region"].unique())
```

```python
date_counts = (
    df.groupby("date")
      .size()
)
print(date_counts.describe())
```

```python
print(
    "Number of unique dates:",
    df["date"].nunique()
)
print(
    "Date range:",
    df["date"].min(),
    "to",
    df["date"].max()
)
```

```python
product_history = (
    df.groupby("product_id")["date"]
      .agg(["min", "max", "count"])
      .sort_values("count")
)
product_history.head(10)
```

```python
product_history.head()
```

```python
product_history.tail()
```

```python
store_product_counts = (
    df.groupby(
        ["store_id", "product_id"]
    )
    .size()
    .reset_index(name="records")
)

store_product_counts.head()
```

```python
Q1 = df["units_sold"].quantile(0.25)
Q3 = df["units_sold"].quantile(0.75)
IQR = Q3 - Q1
lower_bound = Q1 - 1.5 * IQR
upper_bound = Q3 + 1.5 * IQR
outliers = df[
    (df["units_sold"] < lower_bound) |
    (df["units_sold"] > upper_bound)
]
print("Number of potential sales outliers:", len(outliers))
```

```python
df["year"] = df["date"].dt.year
df["month"] = df["date"].dt.month
df["day"] = df["date"].dt.day
df["day_of_week"] = df["date"].dt.dayofweek
df["week_of_year"] = df["date"].dt.isocalendar().week.astype(int)
df["is_weekend"] = (
    df["day_of_week"] >= 5
).astype(int)
```

```python
df[
    [
        "date",
        "year",
        "month",
        "day_of_week",
        "week_of_year",
        "is_weekend"
    ]
].head()
```

```python
df = df.sort_values(
    ["product_id", "store_id", "date"]
).reset_index(drop=True)
```

```python
processed_path = "../data/processed/sales_data_cleaned.csv"
df.to_csv(
    processed_path,
    index=False
)
print("Cleaned dataset saved!")
print("Shape:", df.shape)
```
